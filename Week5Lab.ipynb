```python
#Q2a
myNumbers = np.array([ #creating array with student id
    [2,9,1,0,2,3,7],
    [2,9,1,0,2,3,7],
    [2,9,1,0,2,3,7]
])
myNumbers

#for loop to multiply row 2 by 2 and row 3 by 3
for i in range(1,2): #starting at index 1, ending at index 1, end stop is exclusive
    myNumbers[i] = myNumbers[i] * 2

for i in range(2,3): #starting at index 2, ending at index 2
    myNumbers[i] = myNumbers[i] * 3

print(myNumbers)
```

```python
#Q2b
print("Displaying second row of my array using splicing: ")
print(myNumbers[1:2])
```

```python
#Q2c
print("Displaying first and second rows using splicing: ")
print(myNumbers[:2])
```

```python
#Q2d
print("Displaying second and third rows using fancy indexing")
print(myNumbers[[1,2]])
```

```python
#Q2e
print("Displaying all the rows of values but only the last two columns: ")
print(myNumbers[0:, 5:])
```

```python
#Q2f Boolean indexing
print("Printing numbers greater than 10: ")
print(myNumbers > 10)
print()#formatting
print("Showing full array to check")
print(myNumbers)
```
